```python
df.drop("customerID",axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
binary_cols={
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'PaperlessBilling',
    'Churn'
}
```

```python
for col in binary_cols:
  df[col]=le.fit_transform(df[col])
```

```python
df=pd.get_dummies(df,drop_first=True)
```

```python
df.head()
```

```python
df = df.astype(int)
```

```python
plt.figure(figsize=(20,16))
sns.heatmap(df.corr(numeric_only=True),annot=True,fmt=".2f",cmap='coolwarm')
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
df_scaled=sc.fit_transform(df)
df_scaled=pd.DataFrame(df_scaled,columns=df.columns)
df_scaled.head()
```

```python
x=df.drop('Churn',axis=1)
y=df['Churn']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1,stratify=y)
```

```python
x_train_scaled=sc.fit_transform(x_train)
x_test_scaled=sc.transform(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score,classification_report

def modelperformance(y_test,y_pred):
  print('accuracy Score:',accuracy_score(y_test,y_pred))
  print('Confusion Matrix:',confusion_matrix(y_test,y_pred))
  print('Classification Report:',classification_report(y_test,y_pred))
  print('Precision Score:',precision_score(y_test,y_pred))
  print('Recall Score:',recall_score(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
param_grid_lr = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs', 'saga'],
    'penalty': ['l1', 'l2']
}

grid_lr = GridSearchCV(LogisticRegression(max_iter=1000), param_grid_lr, cv=5, scoring='accuracy', n_jobs=-1)
grid_lr.fit(x_train_scaled, y_train)
```

```python
y_pred_lr_grid = grid_lr.predict(x_test_scaled)
modelperformance(y_test, y_pred_lr_grid)
```

```python
cm_lr_grid = confusion_matrix(y_test, y_pred_lr_grid)
plt.figure(figsize=(8,6))
sns.heatmap(cm_lr_grid,
            annot=True,
            fmt='d',
            cmap='Blues',
            cbar=False,
     xticklabels=['Predicted 0','PREDICTED 1'],
      yticklabels=['Actual 0','Actual 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Logistic Regression (Grid Search)')
plt.show()
```
